Section 7: Python Dictionaries

1. What is a Dictionary?
   A dictionary stores data in key-value pairs.

2. Features =
  - Uses curly brackets '{}'
  - Stores key-value parirs
  - Keys must be unique
  - Values can be duplicated
  - Ordered and mutable

3. Creating a Dictionary
   Student = {"name": "Amit" , "age": 23, "marks": 85}

4. Accessing values =
    Use keys to access values.
    Example: student["name"]

5. Dictionary length
   'len()' returns the numbers of key-value pairs.             

```python
student = {"name": "Amit" , "age": 23, "marks": 85 , "city": "Pune"}
print(student)
print(type(student))
print(len(student))
```

key() returns all values

```python
print(student.keys())
```

values() returns all values from a dictionary

```python
print(student.values())
```

items() returns all key-value pairs

```python
print(student.items())
```

update() adds new key-value pairs or changes existing values

```python
student.update({"city": "Pune"})
print(student)
```

copy() creates a copy of a dictionary

```python
new_student = student.copy()
print(new_student)
print(new_student is student)
```

```python
print(new_student == student)
print(new_student is student)
```

clear() removes all key-value pairs from a dictionary

```python
new_student.clear()
print(new_student)
print(len(new_student))
```

setdefault() returns the value of a key . If the key doesn't exist, it adds the key with a default value.

```python
student = {"name": "Amit" , "age": 23}
print(student.setdefault("city", "Pune"))
print(student)
```

fromkeys() creates a new dictionary using given keys and one common value.

```python
keys = ["name" , "age" , "city"]
new_dict = dict.fromkeys(keys , "unknown")
print(new_dict)
```

popitem() removes and returns the last inserted key-value pair:

```python
student = {"name": "Amit" , "age": 23 , "city": "Pune"}
print(student.popitem())
print(student)
```

del = removes a specific key-value pair.

```python
del student["age"]
print(student)
```

```python
employee = {"name": "Rahul" , "age": 25 , "salary": 30000}
print(employee.setdefault("city" , "Pune"))
print(employee.update({"salary": 35000}))
del employee["age"]
print(employee)
```


Dictionary comprehension:


                           1. num: num**2 = key and square value.

                           2. for  num in numbers = loop through numbers.

                           3. if num % 2 == 0 = select only even numbers.

```python
numbers = [1 , 2 , 3 , 4 ,5]
squares = {num: num**2 for num in numbers}
print(squares)
```

```python
numbers = [2 , 4 , 6 , 8]
doubles = {num: num*2 for num in numbers if num % 2 == 0}
print(doubles)
```

```python
numbers = [2 , 4 , 6 , 8 , 10]
greater = {num: num for num in numbers if num > 5}
print(greater)
```

```python
numbers = [1 , 2 , 3 , 4 , 5 , 6 , 7 , 8]
even = {num: num for num in numbers if num % 2 == 0}
print(even)
```

```python
numbers = [1 , 2 , 3 , 4 , 5 , 6]
squares = {num: num**2 for num in numbers if num % 2 == 0}
print(squares)
```
